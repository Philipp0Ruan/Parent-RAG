# Parent RAG — Os Sertões

Roteiro inicial do Desafio RAG. O pipeline ainda não foi implementado.

Este projeto vai responder perguntas sobre **Os Sertões** buscando trechos curtos e recuperando os blocos maiores aos quais eles pertencem. A proposta é localizar informações específicas sem perder o contexto necessário para interpretar o texto.

Fluxo previsto: PDF → blocos pais → trechos filhos → embeddings dos filhos → busca → recuperação dos pais → resposta com fontes

## 1. Ambiente e configuração

Definir e instalar as dependências, selecionar modelos, carregar variáveis de ambiente e registrar versões e parâmetros. Não inserir chaves no notebook.

## 2. PDF e corpus

Baixar o PDF para `data/os-sertoes.pdf`, registrar seu hash e extrair o texto com fonte e páginas. Conferir a limpeza e separar os textos editoriais do corpus da obra.

A execução deve localizar a raiz do repositório independentemente da pasta em que o Jupyter foi aberto.

## 3. Criar pais e filhos

A implementar: Dividir o texto em blocos maiores e subdividir cada bloco em trechos menores, com parent_id, child_id, fonte e páginas.

## 4. Indexar e armazenar

A implementar: Gerar embeddings apenas dos filhos e guardar os pais em um armazenamento consultável por parent_id.

## 5. Recuperar o contexto

A implementar: Buscar os filhos, resolver seus parent_ids, remover pais duplicados preservando a ordem de relevância e limitar o contexto sem perder a ligação com as páginas.

## 6. Prompt e geração

Montar o contexto com fontes e páginas. Responder em português, distinguir interpretações do autor de fatos atuais e indicar quando faltarem evidências. Usar o mesmo prompt base na comparação entre abordagens.

## 7. Perguntas do desafio

Carregar `evaluation/perguntas.json` e executar as cinco perguntas, preservando os identificadores.

- **Q1:** Qual é a visão de Euclides da Cunha sobre o ambiente natural do sertão nordestino e como ele influencia a vida dos habitantes?
- **Q2:** Quais são as principais características da população sertaneja descritas por Euclides da Cunha? Como ele relaciona essas características com o ambiente em que vivem?
- **Q3:** Qual foi o contexto histórico e político que levou à Guerra de Canudos, segundo Euclides da Cunha?
- **Q4:** Como Euclides da Cunha descreve a figura de Antônio Conselheiro e seu papel na Guerra de Canudos?
- **Q5:** Quais são os principais aspectos da crítica social e política presentes em "Os Sertões"? Como esses aspectos refletem a visão do autor sobre o Brasil da época?

## 8. Avaliação e resultados

A consulta deve buscar os filhos e entregar seus pais ao modelo, com a relação entre os dois visível no notebook e sem repetir o mesmo pai no contexto.

Guardar respostas, evidências, páginas, parâmetros, latência e uso de tokens quando disponível em `results/`. Conferir manualmente a sustentação das respostas no PDF.

Vamos comparar com o Naive RAG para verificar se o contexto mais amplo ajuda nas perguntas interpretativas e quanto aumenta o consumo de contexto.
